### Вариант задания № 1 — Система учёта людей


<h2 style="color:DodgerBlue">Описание проекта:</h2>

----

Проект представляет информацию о людях с помощью базового класса `Person` и производных классов `Student`, `Employee` и `Teacher`. Свойства используют геттеры и сеттеры с проверкой значений, объекты создаются через конструкторы и взаимодействуют между собой во время занятия и профориентации.

#### Дополнительное задание
Добавьте к сущестующим классам конструктора классов с использованием гетторов и сетторов и реализуйте взаимодействие объектов между собой

<h2 style="color:DodgerBlue">Реализация:</h2>

----

```
public class Person
{
    private string _name;
    private int _age;
    private string _gender;

    public string Name
    {
        get => _name;
        set => _name = string.IsNullOrWhiteSpace(value)
            ? throw new ArgumentException("Имя не может быть пустым.")
            : value;
    }

    public int Age
    {
        get => _age;
        set => _age = value >= 0
            ? value
            : throw new ArgumentOutOfRangeException(nameof(value));
    }

    public string Gender
    {
        get => _gender;
        set => _gender = string.IsNullOrWhiteSpace(value)
            ? throw new ArgumentException("Пол должен быть указан.")
            : value;
    }

    public Person(string name, int age, string gender)
    {
        Name = name;
        Age = age;
        Gender = gender;
    }

    public virtual string GetInfo() =>
        $"Имя: {Name}, возраст: {Age}, пол: {Gender}";

    public virtual void SayHello() =>
        Console.WriteLine($"Здравствуйте! Меня зовут {Name}.");
}

public class Student : Person
{
    private string _university;
    private int _course;

    public string University
    {
        get => _university;
        set => _university = string.IsNullOrWhiteSpace(value)
            ? throw new ArgumentException("Университет должен быть указан.")
            : value;
    }

    public int Course
    {
        get => _course;
        set => _course = value is >= 1 and <= 6
            ? value
            : throw new ArgumentOutOfRangeException(nameof(value));
    }

    public Student(string name, int age, string gender, string university, int course)
        : base(name, age, gender)
    {
        University = university;
        Course = course;
    }

    public override string GetInfo() =>
        $"{base.GetInfo()}, университет: {University}, курс: {Course}";

    public override void SayHello() =>
        Console.WriteLine($"Здравствуйте! Меня зовут {Name}, я учусь на {Course} курсе.");

    public void AttendLesson(Teacher teacher) =>
        Console.WriteLine($"{Name} посещает занятие по предмету «{teacher.Subject}».");
}

public class Employee : Person
{
    private string _company;
    private decimal _salary;

    public string Company
    {
        get => _company;
        set => _company = string.IsNullOrWhiteSpace(value)
            ? throw new ArgumentException("Компания должна быть указана.")
            : value;
    }

    public decimal Salary
    {
        get => _salary;
        set => _salary = value >= 0
            ? value
            : throw new ArgumentOutOfRangeException(nameof(value));
    }

    public Employee(string name, int age, string gender, string company, decimal salary)
        : base(name, age, gender)
    {
        Company = company;
        Salary = salary;
    }

    public override string GetInfo() =>
        $"{base.GetInfo()}, компания: {Company}, зарплата: {Salary:F2} руб.";

    public void GiveCareerAdvice(Student student) =>
        Console.WriteLine($"{Name} рассказывает студенту {student.Name} о работе в компании «{Company}».");
}

public class Teacher : Person
{
    private string _subject;
    private int _experience;

    public string Subject
    {
        get => _subject;
        set => _subject = string.IsNullOrWhiteSpace(value)
            ? throw new ArgumentException("Предмет должен быть указан.")
            : value;
    }

    public int Experience
    {
        get => _experience;
        set => _experience = value >= 0
            ? value
            : throw new ArgumentOutOfRangeException(nameof(value));
    }

    public Teacher(string name, int age, string gender, string subject, int experience)
        : base(name, age, gender)
    {
        Subject = subject;
        Experience = experience;
    }

    public override string GetInfo() =>
        $"{base.GetInfo()}, предмет: {Subject}, стаж: {Experience} лет";

    public override void SayHello() =>
        Console.WriteLine($"Здравствуйте! Меня зовут {Name}, я преподаю предмет «{Subject}».");

    public void Teach(Student student) =>
        Console.WriteLine($"{Name} проводит занятие для студента {student.Name}.");
}

var student = new Student("Максим", 20, "мужской", "ТИУ", 2);
var employee = new Employee("Анна", 28, "женский", "ТехноСофт", 90000m);
var teacher = new Teacher("Иван", 40, "мужской", "ООП", 12);

Person[] people = { student, employee, teacher };
foreach (Person person in people)
{
    Console.WriteLine(person.GetInfo());
    person.SayHello();
}

teacher.Teach(student);
student.AttendLesson(teacher);
employee.GiveCareerAdvice(student);
```

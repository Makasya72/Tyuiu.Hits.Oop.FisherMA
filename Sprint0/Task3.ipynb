### Вариант задания № 1 — Система учёта людей


<h2 style="color:DodgerBlue">Описание проекта:</h2>

----

Проект представляет информацию о людях с помощью базового класса `Person`. Производные классы `Student`, `Employee` и `Teacher` расширяют его данными об учёбе и работе. Наследование и переопределение методов `GetInfo()` и `SayHello()` демонстрируют полиморфизм.


<h2 style="color:DodgerBlue">Реализация:</h2>

----

```
public class Person
{
    public string Name { get; set; }
    public int Age { get; set; }
    public string Gender { get; set; }

    public Person(string name, int age, string gender)
    {
        Name = name;
        Age = age;
        Gender = gender;
    }

    public virtual string GetInfo() =>
        $"Имя: {Name}, возраст: {Age}, пол: {Gender}";

    public virtual void SayHello() =>
        Console.WriteLine($"Здравствуйте! Меня зовут {Name}.");
}

public class Student : Person
{
    public string University { get; set; }
    public int Course { get; set; }

    public Student(string name, int age, string gender, string university, int course)
        : base(name, age, gender)
    {
        University = university;
        Course = course;
    }

    public override string GetInfo() =>
        $"{base.GetInfo()}, университет: {University}, курс: {Course}";

    public override void SayHello() =>
        Console.WriteLine($"Здравствуйте! Меня зовут {Name}, я учусь на {Course} курсе.");
}

public class Employee : Person
{
    public string Company { get; set; }
    public decimal Salary { get; set; }

    public Employee(string name, int age, string gender, string company, decimal salary)
        : base(name, age, gender)
    {
        Company = company;
        Salary = salary;
    }

    public override string GetInfo() =>
        $"{base.GetInfo()}, компания: {Company}, зарплата: {Salary:F2} руб.";
}

public class Teacher : Person
{
    public string Subject { get; set; }
    public int Experience { get; set; }

    public Teacher(string name, int age, string gender, string subject, int experience)
        : base(name, age, gender)
    {
        Subject = subject;
        Experience = experience;
    }

    public override string GetInfo() =>
        $"{base.GetInfo()}, предмет: {Subject}, стаж: {Experience} лет";

    public override void SayHello() =>
        Console.WriteLine($"Здравствуйте! Меня зовут {Name}, я преподаю предмет «{Subject}».");
}

Person[] people =
{
    new Student("Максим", 20, "мужской", "ТИУ", 2),
    new Employee("Анна", 28, "женский", "ТехноСофт", 90000m),
    new Teacher("Иван", 40, "мужской", "ООП", 12)
};

foreach (Person person in people)
{
    Console.WriteLine(person.GetInfo());
    person.SayHello();
}
```

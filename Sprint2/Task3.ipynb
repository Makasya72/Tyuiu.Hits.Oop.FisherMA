### Вариант задания № 1 — Система учёта людей


<h2 style="color:DodgerBlue">Описание проекта:</h2>

----

Проект построен вокруг базового класса `Person`. Классы `Student` и `Employee` демонстрируют простое наследование, цепочка `Person → Employee → Teacher` — сложное многоуровневое наследование, а реализация преподавателем интерфейсов `ITeaching` и `IResearching` — множественное наследование по контрактам. Каждый класс содержит собственные атрибуты, методы и проверяемые свойства.

#### Дополнительное задание
Добавьте к сущестующим классам (базовыму и производным 3-4 атрибута и метода) и реализуйте простое, сложное и множественное наследование

<h2 style="color:DodgerBlue">Реализация:</h2>

----

```
public interface ITeaching
{
    void Teach(Student student);
    void GradeStudent(Student student, double grade);
}

public interface IResearching
{
    string ResearchTopic { get; set; }
    void ConductResearch();
}

public class Person
{
    private string _name;
    private int _age;
    private string _gender;
    private string _email;

    public string Name
    {
        get => _name;
        set => _name = string.IsNullOrWhiteSpace(value)
            ? throw new ArgumentException("Имя не может быть пустым.")
            : value;
    }

    public int Age
    {
        get => _age;
        set => _age = value >= 0
            ? value
            : throw new ArgumentOutOfRangeException(nameof(value));
    }

    public string Gender
    {
        get => _gender;
        set => _gender = string.IsNullOrWhiteSpace(value) ? "не указан" : value;
    }

    public string Email
    {
        get => _email;
        set => _email = !string.IsNullOrWhiteSpace(value) && value.Contains('@')
            ? value
            : throw new ArgumentException("Некорректный адрес электронной почты.");
    }

    public Person(string name, int age, string gender, string email)
    {
        Name = name;
        Age = age;
        Gender = gender;
        Email = email;
    }

    public virtual string GetInfo() =>
        $"Имя: {Name}, возраст: {Age}, пол: {Gender}, e-mail: {Email}";

    public virtual void SayHello() =>
        Console.WriteLine($"Здравствуйте! Меня зовут {Name}.");

    public void HaveBirthday() => Age++;
    public void ChangeEmail(string email) => Email = email;
}

public class Student : Person
{
    public string University { get; set; }
    public int Course { get; private set; }
    public string Group { get; set; }
    public double AverageGrade { get; private set; }

    public Student(string name, int age, string gender, string email,
        string university, int course, string group, double averageGrade)
        : base(name, age, gender, email)
    {
        University = university;
        Course = Math.Clamp(course, 1, 6);
        Group = group;
        AverageGrade = Math.Clamp(averageGrade, 2, 5);
    }

    public override string GetInfo() =>
        $"{base.GetInfo()}, университет: {University}, курс: {Course}, группа: {Group}, средний балл: {AverageGrade:F2}";

    public override void SayHello() =>
        Console.WriteLine($"Здравствуйте! Я {Name}, студент {Course} курса {University}.");

    public void Study() => Console.WriteLine($"{Name} изучает материал.");
    public void TakeExam(double grade) => AverageGrade = Math.Clamp((AverageGrade + grade) / 2, 2, 5);
    public void AdvanceCourse() => Course = Math.Min(6, Course + 1);
}

public class Employee : Person
{
    public string Company { get; set; }
    public string Position { get; private set; }
    public decimal Salary { get; private set; }
    public int YearsOfService { get; protected set; }

    public Employee(string name, int age, string gender, string email,
        string company, string position, decimal salary, int yearsOfService)
        : base(name, age, gender, email)
    {
        Company = company;
        Position = position;
        Salary = Math.Max(0, salary);
        YearsOfService = Math.Max(0, yearsOfService);
    }

    public override string GetInfo() =>
        $"{base.GetInfo()}, компания: {Company}, должность: {Position}, зарплата: {Salary:F2} руб., стаж: {YearsOfService} лет";

    public void Work() => Console.WriteLine($"{Name} работает в компании «{Company}».");
    public void ReceiveBonus(decimal amount) => Salary += Math.Max(0, amount);
    public void Promote(string position, decimal salaryIncrease)
    {
        Position = position;
        Salary += Math.Max(0, salaryIncrease);
    }
}

public class Teacher : Employee, ITeaching, IResearching
{
    public string Subject { get; set; }
    public int Experience { get; set; }
    public string AcademicDegree { get; set; }
    public string ResearchTopic { get; set; }

    public Teacher(string name, int age, string gender, string email,
        string company, decimal salary, int yearsOfService, string subject,
        int experience, string academicDegree, string researchTopic)
        : base(name, age, gender, email, company, "преподаватель", salary, yearsOfService)
    {
        Subject = subject;
        Experience = Math.Max(0, experience);
        AcademicDegree = academicDegree;
        ResearchTopic = researchTopic;
    }

    public override string GetInfo() =>
        $"{base.GetInfo()}, предмет: {Subject}, преподавательский стаж: {Experience} лет, степень: {AcademicDegree}";

    public override void SayHello() =>
        Console.WriteLine($"Здравствуйте! Я {Name}, преподаю предмет «{Subject}».");

    public void Teach(Student student) =>
        Console.WriteLine($"{Name} проводит занятие по предмету «{Subject}» для студента {student.Name}.");

    public void GradeStudent(Student student, double grade)
    {
        student.TakeExam(grade);
        Console.WriteLine($"{Name} поставил студенту {student.Name} оценку {grade:F1}.");
    }

    public void ConductResearch() =>
        Console.WriteLine($"{Name} проводит исследование по теме «{ResearchTopic}».");
}

var student = new Student("Максим", 20, "мужской", "maksim@example.com",
    "ТИУ", 2, "ИСТб-24-1", 4.5);
var employee = new Employee("Анна", 28, "женский", "anna@example.com",
    "ТехноСофт", "разработчик", 90000m, 5);
var teacher = new Teacher("Иван", 40, "мужской", "ivan@example.com",
    "ТИУ", 110000m, 12, "ООП", 12, "кандидат технических наук", "Архитектура программных систем");

Person[] people = { student, employee, teacher };
foreach (Person person in people)
{
    Console.WriteLine(person.GetInfo());
    person.SayHello();
}

student.Study();
employee.Work();
teacher.Teach(student);
teacher.GradeStudent(student, 5);
teacher.ConductResearch();
Console.WriteLine($"Новый средний балл студента: {student.AverageGrade:F2}.");
```

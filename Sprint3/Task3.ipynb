### Вариант задания № 1 — Система учёта людей


<h2 style="color:DodgerBlue">Описание проекта:</h2>

----

Проект использует базовый класс `Person` и производные классы `Student`, `Employee` и `Teacher`. Переопределение методов обеспечивает полиморфное поведение, перегрузка позволяет выполнять операции с разным набором параметров, а Generic-класс `PersonRegistry<T>` хранит и обрабатывает объекты заданного типа.

#### Дополнительное задание
Добавьте к сущестующим классам (базовыму и производным 3-4 атрибута и метода) и реализуйте полиморфизм с перекрытием и прегегрузкой методов, а также generic классы

<h2 style="color:DodgerBlue">Реализация:</h2>

----

```
public class Person
{
    public string Name { get; set; }
    public int Age { get; set; }
    public string Gender { get; set; }
    public string Email { get; set; }

    public Person(string name, int age, string gender, string email)
    {
        Name = name;
        Age = Math.Max(0, age);
        Gender = gender;
        Email = email;
    }

    public virtual string GetInfo() =>
        $"Имя: {Name}, возраст: {Age}, пол: {Gender}, e-mail: {Email}";

    public virtual void SayHello() => Console.WriteLine($"Здравствуйте! Меня зовут {Name}.");

    public void UpdateContact(string email) => Email = email;
    public void UpdateContact(string email, string message) =>
        Console.WriteLine($"{Name}: {message}. Новый e-mail: {Email = email}");
}

public class Student : Person
{
    public string University { get; set; }
    public int Course { get; set; }
    public string Group { get; set; }
    public double AverageGrade { get; private set; }

    public Student(string name, int age, string gender, string email,
        string university, int course, string group, double averageGrade)
        : base(name, age, gender, email)
    {
        University = university;
        Course = course;
        Group = group;
        AverageGrade = averageGrade;
    }

    public override string GetInfo() =>
        $"{base.GetInfo()}, университет: {University}, курс: {Course}, группа: {Group}, средний балл: {AverageGrade:F2}";

    public override void SayHello() =>
        Console.WriteLine($"Здравствуйте! Я {Name}, студент {Course} курса {University}.");

    public void TakeExam(double grade) => AverageGrade = Math.Clamp((AverageGrade + grade) / 2, 2, 5);
    public void TakeExam(string subject, double grade)
    {
        TakeExam(grade);
        Console.WriteLine($"{Name} сдал предмет «{subject}» на {grade:F1}.");
    }
}

public class Employee : Person
{
    public string Company { get; set; }
    public string Position { get; set; }
    public decimal Salary { get; private set; }
    public int YearsOfService { get; set; }

    public Employee(string name, int age, string gender, string email,
        string company, string position, decimal salary, int yearsOfService)
        : base(name, age, gender, email)
    {
        Company = company;
        Position = position;
        Salary = Math.Max(0, salary);
        YearsOfService = Math.Max(0, yearsOfService);
    }

    public override string GetInfo() =>
        $"{base.GetInfo()}, компания: {Company}, должность: {Position}, зарплата: {Salary:F2} руб., стаж: {YearsOfService} лет";

    public void ReceiveBonus(decimal amount) => Salary += Math.Max(0, amount);
    public void ReceiveBonus(decimal amount, string reason)
    {
        ReceiveBonus(amount);
        Console.WriteLine($"{Name} получил премию {amount:F2} руб. Причина: {reason}.");
    }
}

public class Teacher : Employee
{
    public string Subject { get; set; }
    public int Experience { get; set; }
    public string AcademicDegree { get; set; }
    public string ResearchTopic { get; set; }

    public Teacher(string name, int age, string gender, string email,
        string company, decimal salary, int yearsOfService, string subject,
        int experience, string academicDegree, string researchTopic)
        : base(name, age, gender, email, company, "преподаватель", salary, yearsOfService)
    {
        Subject = subject;
        Experience = experience;
        AcademicDegree = academicDegree;
        ResearchTopic = researchTopic;
    }

    public override string GetInfo() =>
        $"{base.GetInfo()}, предмет: {Subject}, степень: {AcademicDegree}, тема исследования: {ResearchTopic}";

    public override void SayHello() =>
        Console.WriteLine($"Здравствуйте! Я {Name}, преподаю предмет «{Subject}».");

    public void Teach(Student student) =>
        Console.WriteLine($"{Name} проводит занятие для студента {student.Name}.");

    public void Teach(IEnumerable<Student> students) =>
        Console.WriteLine($"{Name} проводит занятие для группы из {students.Count()} студентов.");
}

public class PersonRegistry<T> where T : Person
{
    private readonly List<T> _people = new();

    public int Count => _people.Count;
    public void Add(T person) => _people.Add(person);
    public T FindByName(string name) => _people.FirstOrDefault(p => p.Name == name);

    public void PrintAll()
    {
        foreach (T person in _people)
            Console.WriteLine(person.GetInfo());
    }
}

var student = new Student("Максим", 20, "мужской", "maksim@example.com",
    "ТИУ", 2, "ИСТб-24-1", 4.5);
var employee = new Employee("Анна", 28, "женский", "anna@example.com",
    "ТехноСофт", "разработчик", 90000m, 5);
var teacher = new Teacher("Иван", 40, "мужской", "ivan@example.com",
    "ТИУ", 110000m, 12, "ООП", 12, "кандидат технических наук",
    "Архитектура программных систем");

var registry = new PersonRegistry<Person>();
registry.Add(student);
registry.Add(employee);
registry.Add(teacher);
registry.PrintAll();

Person[] people = { student, employee, teacher };
foreach (Person person in people)
    person.SayHello();

student.TakeExam("ООП", 5);
employee.ReceiveBonus(10000m, "успешный проект");
teacher.Teach(student);
teacher.Teach(new[] { student });
Console.WriteLine($"В реестре: {registry.Count} человека.");
```

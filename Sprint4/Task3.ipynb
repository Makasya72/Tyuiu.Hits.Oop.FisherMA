### Вариант задания № 1 — Система учёта людей


<h2 style="color:DodgerBlue">Описание проекта:</h2>

----

Проект использует базовый класс `Person` и производные классы `Student`, `Employee` и `Teacher`. Интерфейс `IPrintable` реализован явно, поэтому доступен только через контракт. Специализированные действия ролей также реализованы явно. Класс `PersonManager<T>` получает сервис уведомлений через конструктор, демонстрируя управление зависимостями.

#### Дополнительное задание
Добавьте к сущестующим классам (базовыму и производным 3-4 атрибута и метода) создайте явную реализации интерфейса и управление зависимостями 


<h2 style="color:DodgerBlue">Реализация:</h2>

----

```
public interface IPrintable
{
    string GetDescription();
}

public interface IStudyActivity
{
    void Perform();
}

public interface IWorkActivity
{
    void Perform();
}

public interface ITeachingActivity
{
    void Perform(Student student);
}

public interface INotificationService
{
    void Send(Person person, string message);
}

public class ConsoleNotificationService : INotificationService
{
    public void Send(Person person, string message) =>
        Console.WriteLine($"Уведомление для {person.Name}: {message}");
}

public class Person : IPrintable
{
    public string Name { get; set; }
    public int Age { get; set; }
    public string Gender { get; set; }
    public string Email { get; set; }

    public Person(string name, int age, string gender, string email)
    {
        Name = name;
        Age = Math.Max(0, age);
        Gender = gender;
        Email = email;
    }

    public virtual string GetInfo() =>
        $"Имя: {Name}, возраст: {Age}, пол: {Gender}, e-mail: {Email}";

    public virtual void SayHello() => Console.WriteLine($"Здравствуйте! Меня зовут {Name}.");
    public void HaveBirthday() => Age++;
    public void ChangeEmail(string email) => Email = email;

    string IPrintable.GetDescription() => GetInfo();
}

public class Student : Person, IStudyActivity
{
    public string University { get; set; }
    public int Course { get; set; }
    public string Group { get; set; }
    public double AverageGrade { get; set; }

    public Student(string name, int age, string gender, string email,
        string university, int course, string group, double averageGrade)
        : base(name, age, gender, email)
    {
        University = university;
        Course = course;
        Group = group;
        AverageGrade = averageGrade;
    }

    public override string GetInfo() =>
        $"{base.GetInfo()}, университет: {University}, курс: {Course}, группа: {Group}, средний балл: {AverageGrade:F2}";

    public override void SayHello() => Console.WriteLine($"Здравствуйте! Я {Name}, студент {Course} курса {University}.");

    void IStudyActivity.Perform() => Console.WriteLine($"{Name} выполняет учебное задание.");
}

public class Employee : Person, IWorkActivity
{
    public string Company { get; set; }
    public string Position { get; set; }
    public decimal Salary { get; set; }
    public int YearsOfService { get; set; }

    public Employee(string name, int age, string gender, string email,
        string company, string position, decimal salary, int yearsOfService)
        : base(name, age, gender, email)
    {
        Company = company;
        Position = position;
        Salary = salary;
        YearsOfService = yearsOfService;
    }

    public override string GetInfo() =>
        $"{base.GetInfo()}, компания: {Company}, должность: {Position}, зарплата: {Salary:F2} руб., стаж: {YearsOfService} лет";

    void IWorkActivity.Perform() => Console.WriteLine($"{Name} работает в компании «{Company}».");
}

public class Teacher : Employee, ITeachingActivity
{
    public string Subject { get; set; }
    public int Experience { get; set; }
    public string AcademicDegree { get; set; }
    public string ResearchTopic { get; set; }

    public Teacher(string name, int age, string gender, string email,
        string company, decimal salary, int yearsOfService, string subject,
        int experience, string academicDegree, string researchTopic)
        : base(name, age, gender, email, company, "преподаватель", salary, yearsOfService)
    {
        Subject = subject;
        Experience = experience;
        AcademicDegree = academicDegree;
        ResearchTopic = researchTopic;
    }

    public override string GetInfo() =>
        $"{base.GetInfo()}, предмет: {Subject}, степень: {AcademicDegree}, исследование: {ResearchTopic}";

    public override void SayHello() => Console.WriteLine($"Здравствуйте! Я {Name}, преподаю предмет «{Subject}».");

    void ITeachingActivity.Perform(Student student) =>
        Console.WriteLine($"{Name} проводит занятие по предмету «{Subject}» для студента {student.Name}.");
}

public class PersonManager<T> where T : Person
{
    private readonly INotificationService _notificationService;
    private readonly List<T> _people = new();

    public PersonManager(INotificationService notificationService) =>
        _notificationService = notificationService;

    public void Add(T person)
    {
        _people.Add(person);
        _notificationService.Send(person, "вы добавлены в систему");
    }

    public void NotifyAll(string message)
    {
        foreach (T person in _people)
            _notificationService.Send(person, message);
    }

    public void PrintAll()
    {
        foreach (T person in _people)
            Console.WriteLine(((IPrintable)person).GetDescription());
    }
}

var student = new Student("Максим", 20, "мужской", "maksim@example.com",
    "ТИУ", 2, "ИСТб-24-1", 4.5);
var employee = new Employee("Анна", 28, "женский", "anna@example.com",
    "ТехноСофт", "разработчик", 90000m, 5);
var teacher = new Teacher("Иван", 40, "мужской", "ivan@example.com",
    "ТИУ", 110000m, 12, "ООП", 12, "кандидат технических наук",
    "Архитектура программных систем");

INotificationService notificationService = new ConsoleNotificationService();
var manager = new PersonManager<Person>(notificationService);
manager.Add(student);
manager.Add(employee);
manager.Add(teacher);
manager.PrintAll();

((IStudyActivity)student).Perform();
((IWorkActivity)employee).Perform();
((ITeachingActivity)teacher).Perform(student);
manager.NotifyAll("занятие начинается в 10:00");
```
